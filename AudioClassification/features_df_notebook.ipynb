```python
import pandas as pd
from pathlib import Path

def widen_single_label_df(df_single, idx_to_class, label_col="label"):
    """
    Convert a single-label df (columns: filepath, label, <features...>) into
    the same wide format as the base+overlay dataset:
      - label_base: renamed from the original label column (the base
        audio's own class code, unchanged)
      - label_overlay_<class>: one one-hot column per class, 1 only for the
        row's own class.

    idx_to_class : dict[int, str]
        Maps each label code -> class name, matching the naming used in
        your original df's label_overlay_<class> columns. Must use the
        SAME encoding as was used to build the base+overlay df.
    """
    df = df_single.copy()
    df = df.rename(columns={label_col: "label_base"})

    unknown = set(df["label_base"].unique()) - set(idx_to_class.keys())
    if unknown:
        raise ValueError(f"label contains codes not in idx_to_class: {unknown}")

    class_names = [idx_to_class[i] for i in sorted(idx_to_class)]
    for name in class_names:
        df[f"label_overlay_{name}"] = 0

    for idx, name in idx_to_class.items():
        df.loc[df["label_base"] == idx, f"label_overlay_{name}"] = 1

    return df


# --- usage ---
IDX_TO_CLASS = {
    0: "Airplanes",
    1: "Drones",
    2: "Explosion",
    3: "Gunshots",
    4: "Helicopter",
    5: "Land_Vehicles",
    6: "shortened_unlabeled",
}

df_single = pd.read_csv(Path(r".\Data_v2\Bootstrapped_Audios\FeatureTable\test_df.csv"))
df_base_overlay = pd.read_csv(Path(r".\Data_v3\Bootstrapped_Audios\FeatureTable\test_df.csv"))

df_single_wide = widen_single_label_df(df_single, IDX_TO_CLASS, label_col="label")
df_combined = pd.concat([df_base_overlay, df_single_wide], axis=0, ignore_index=True)
```

```python
df_combined.to_csv("test_df.csv", index=False)
```
